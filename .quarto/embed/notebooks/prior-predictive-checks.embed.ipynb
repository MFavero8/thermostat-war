## Sex-specific intercepts

The model in <a href="#eq-1">Equation (1)</a> is simplified to a model with a sex-specific intercepts. Specifically:

$$
\begin{aligned}
\mathbf Y_i &\sim \operatorname{Multinomial}(5,\mathbf p_i)
\\[6pt]
\mathbf p_i &= \operatorname{Softmax}(\boldsymbol\eta_i)
\\[6pt]
\boldsymbol\eta_i &=
(\eta_{i,-3},\eta_{i,-2},\eta_{i,-1},0,
\eta_{i,1},\eta_{i,2},\eta_{i,3})^{\mathsf T}
\\[6pt]
\eta_{i,k} &=
\bar{\alpha}_{\mathrm{sex}[i],k}
\end{aligned}
$$
